## os module
- this python module enables interacting with host operating system

```python
import os
```

###  primary methods
- os.getcwd() `returns directory`
- os.listdir() `returns list of files in dir`
- os.mkdir() `creates folder in dir`
- os.rmdir() `removes the empty dir`
- os.remove() `removes the file specified`
- os.chdir() `changes the current dir to specified one`

```python
print(os.getcwd())
```

```python
print(os.listdir())
```

```python
os.mkdir('R')
```

```python
os.rmdir('R')
```
