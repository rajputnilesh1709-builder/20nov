### Exception Handling
- Exception are events that can modify the flow of program
- in python the runtime exceptions are handled using following
   1.try- except
  2.try-finally
  3.raise
  4.assert

### try-except
**syntax**
```python
try:
    statements
except name1:
    this will run if name1 exception occurs
except name2,name3:
    this will run if any of these exceptions occur
except name4 as var:
    this will run if name4 exception occurs and var is assigned to it
except :
    this will run for other exceptions
else:
    this will run if no exception occurs
    
    
    

````python
try:
    print("h"+3)
except :
    print('error occurs')
else:

    print('success')
````

````python
try:
    print(1/0)
except ValueError:
    print('error occurs')
except Exception as e:
    print('other')
    print(e)
    print(type(e))
else:

    print('success')
````
